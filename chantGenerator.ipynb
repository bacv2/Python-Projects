```python
while True:
    try:
        ralphie_score = int(input("Enter Ralphie's score: "))
        chip_score = int(input("Enter Chip's score: "))

        #apply error to negative numbers
        if ralphie_score <= 0 or chip_score <= 0:
            raise ValueError
    #error if input isn't a positive number        
    except ValueError:
        print("Score must be a positive number!")
        continue
    #break to avoid infinite loop
    break
#Newline character for spacing
print("\n")
#If the scores are equal, announce a tie. Otherwise, determine the winner by who has the higher trivia score.
if ralphie_score == chip_score:
    print("it's a tie!")
#They only get something if one of them wins so the tie conditional is only the print statement.
elif ralphie_score > chip_score:
    print("The scores are different.")
    print("Ralphie wins!")
    #Check each player's score
    if ralphie_score >= 10:
        print("Ralphie earned a prize✨")
    if ralphie_score >= 4 and ralphie_score <= 10:
        print('Ralphie gets a high five!')
elif chip_score > ralphie_score:
    print("The scores are different.")
    print("Chip wins!")
    #Check each player's score
    if chip_score >= 10:
        print("Chip earned a prize✨")
    if chip_score >= 4 and chip_score <= 10:
        print('Chip gets a high five!')

#Seperate if-nests to handle bad scores.
if ralphie_score <= 3 or chip_score <= 3:
    if ralphie_score <=3:
        print("Ralphie must complete a practice round.")
    if chip_score <=3:
        print("Chip must complete a practice round.")       
```
