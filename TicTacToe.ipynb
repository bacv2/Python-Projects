```python
class TicTacToe:
    def __init__(self):
        self.board = []
    
    def createBoard(self, height=3, width=3):
        for num in range(width):
            self.board.append(list('.'*height))
        return self.board

    def printBoard(self):
        #source: GeeksforGeeks
        for items in self.board:
            for i in items:
                print(i, end='  ')
            print()
        print()
        return self.board

    def placePiece(self, row, column, piece):
        self.board[row][column] = piece
        self.printBoard()
    
game = TicTacToe()
game.createBoard()
game.printBoard()

END = 'END'
playerInput = ""

while playerInput != END:
    playerInput = input('Enter "row, col, piece" or "END":')
    if playerInput == END:
        break
    try:
        #source: GeeksforGeeks
        row, column, piece = playerInput.split(',')
        game.placePiece(int(row), int(column), str(piece))       
    except (ValueError, IndexError):
        print("Invalid input or coordinates. Please use format row, column, piece.")
```
