```python
while True:
    try:
        number_of_sentences = int(input(('Please input the number of sentences: ')))
        number_of_anotators = int(input(('Please input the number of anotators: ')))
    except ValueError:
        print('Input value must be an integer!')
        continue

    if number_of_sentences <= 0 or number_of_anotators <= 0:
        raise ValueError('Please enter a positive integer.')
        continue
    break

exact_average = number_of_sentences / number_of_anotators
complete_sentence_average = number_of_sentences // number_of_anotators
left_over_average = number_of_sentences % number_of_anotators

print(f'Number of sentences: {number_of_sentences}')
print(f'Number of anotators: {number_of_anotators}', '\n')
print(f'Exact average: {exact_average} sentences per anotator')
print(f'Complete sentences assigned per annotator: {complete_sentence_average}')
print(f'Sentences left over: {left_over_average}')
```
