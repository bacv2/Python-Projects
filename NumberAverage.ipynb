```python
#Asking for numbers
#try keyword: allows to test code for errors.
try:
    user_input1 = float(input('Enter a number:'))
    #output: Enter a number: (asks for user input)
    user_input2 = float(input('Enter another number:'))
    #output: Enter another number: (asks for user input)
    #Printing out average
    print(f'The average is: {(user_input1 + user_input2)/2}')
    #output: The average is: (average)
#except keyword: lets you handle the error and apply a custom message.
except ValueError:
    print('Please enter a number!')
#Applied exception to handle non-valid user inputs
#Source: Consulted w3schools to review exception handling (https://www.w3schools.com/python/python_try_except.asp)
```
